```python
import os
image_filenames = list(filter(lambda x: "jpeg" in x.lower(), os.listdir()))
```

```python
# Data Never Sleeps Infographuc 2011-2022
# https://pillow.readthedocs.io/en/stable/
from PIL import Image

Image.open(image_filenames[0])
```

```python
# Homogeneizar imágenes
n = 2
img = Image.open(image_filenames[n]).convert('RGB')
pic = np.asarray(img)
fig, axs = plt.subplots(figsize=(10, 10))
plt.imshow(pic[:,:])
plt.show()
```

```python
image_filenames.sort()
image_filenames.remove('pic10.jpeg',)
image_filenames.remove('pic11.jpeg',)
image_filenames += ['pic10.jpeg', 'pic11.jpeg']
image_filenames
```

```python
v_min=[175, 450, 550, 350, 300, 0, 200, 300, 400, 550, 950]
v_max=[1275, 3250, 2400, 1550, 1300, 1000, 875, 1300, 1600, 2200, 3450]
users=[2.1, 2.4, 3.2, 3.4, 3.7, 3.8, 4.3, 4.5, 5.0, 5.1, 5.2]
```

```python
images = [Image.open(fname).convert('RGB') for fname in image_filenames]
images = [np.asarray(img)[v_min[n]:v_max[n], :, :] for n, img in enumerate(images)]
images = [Image.fromarray(img).resize((1000, 1000), Image.LANCZOS) for img in images]
```

```python
np.asarray(images[0]).shape, np.asarray(images[-1]).shape
```

```python
n = 1
pic = np.asarray(images[n])
fig, axs = plt.subplots(figsize=(4, 6))
plt.imshow(pic[:, :])
plt.axis("off")
plt.title("Data Never Sleeps, {}".format(2011+n), fontsize="16", fontweight="bold")
plt.show()
```

```python
for n in range(len(images)):
  pic = np.asarray(images[n])
  fig, axs = plt.subplots(figsize=(4, 6))
  plt.imshow(pic[:, :])
  plt.axis("off")
  plt.title("Data Never Sleeps, {}".format(2011+n), fontsize="16", fontweight="bold")
  plt.savefig("DNS_{}".format(2011+n))
```

```python
frames = [Image.open('DNS_{}.png'.format(2011 + n)).convert('RGB')
          for n in range(len(images))]
```

```python
url_wiki = "https://es.wikipedia.org/wiki/Primera_Divisi%C3%B3n_de_M%C3%A9xico"
```

```python
len(pd.read_html(url_wiki))
```

```python
wiki_page = pd.read_html(url_wiki)
df = wiki_page[4]
df.head()
```

```python
df.columns = ['Club', 'Títulos', 'Subtítulos', 'Campeonatos',
       'Años subcampeón']
df = df[['Club', 'Títulos', 'Campeonatos']]
```

```python
import warnings
warnings.filterwarnings('ignore')

df.dropna(inplace=True)
df.head()
```

```python
df.iloc[-13,2]
```

```python
[year[-2:] for year in df.iloc[-13, 2].replace(' e', ',').replace(' y', ',').split(",")]
```

```python
df['Campeonatos'] = df['Campeonatos'].apply(
    lambda x: [year[-2:] for year in x.replace(
        ' e',',').replace(' y',',').split(',')])
```

```python
df['Campeonatos']=df['Campeonatos'].apply(
    lambda x: [int('19'+year) if int(year)>30 else int('20'+year) for year in x])
```

```python
# Pandas Explode
# https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.explode.html

time_line = df.explode('Campeonatos').reset_index()
time_line.sort_values('Campeonatos')
```

```python
dummies = pd.get_dummies(time_line['Campeonatos'], dtype=int)
```

```python
final = pd.concat([time_line[["Club"]], dummies], axis=1)
final = final.groupby('Club').sum()
final = final.cumsum(axis=1)
final
```

```python
final.head(10)
```

```python
year = 2025
sns.barplot(data=final.sort_values(year, ascending=False).head(10)[year], orient='h', palette='rocket')
plt.text(final[year].max(), 9, year, ha='right')
plt.xlim(0, final[year].max())
plt.title("Campeones primera división Futbol MX")
plt.show()
```

```python
def Bar_Char_race(year):
  axs.clear()
  sns.barplot(data=final.sort_values(year, ascending=False).head(10)[year], orient='h', palette='rocket', ax=axs)
  axs.text(final[year].max(), 9, year, ha='right')
  axs.set_xlim(0, final[year].max())
  axs.set_title("Campeones primera división Futbol MX")
  axs.set_xlabel("Número de Campeonatos")
```

```python
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

# fig, axs = plt.subplots()
movie = FuncAnimation(fig, Bar_Char_race, frames=final.columns)
HTML(movie.to_jshtml())
```
