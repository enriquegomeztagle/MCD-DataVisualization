```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from textwrap import wrap
from matplotlib import font_manager

font_path = "/Users/<YOUR_USER>/Library/Fonts/"
font_name = "AlbertSans-Regular.ttf"
full_font_path = font_path + font_name
font = font_manager.FontProperties(fname=full_font_path)

categories = [
    "Incrementar productividad",
    "Crear contenido ágil",
    "Automatizar tareas",
    "Mejorar retención",
    "Ahorro en costos"
]
percentages = [65.7, 58.6, 38.0, 68.0, 55.0]
colors = ['#c340ff', '#70e0ff', '#5f5ff6', '#A673E0', '#70a0ff']

N = len(categories)
angles = np.linspace(0, 2 * np.pi, N, endpoint=False).tolist()
angles += angles[:1]

fig, ax = plt.subplots(figsize=(8, 8), subplot_kw=dict(polar=True), facecolor='none')
fig.patch.set_alpha(0.0)
fig.suptitle(
    "Beneficios Clave de la IA Generativa",
    fontsize=16,
    y=0.95,
    color='white',
    fontproperties=font
)
fig.subplots_adjust(top=0.8, bottom=0.05, right=0.85, left=0.15)

def animate(frame):
    ax.clear()
    ax.set_ylim(0, 90)
    ax.set_theta_offset(np.pi / 2)
    ax.set_theta_direction(-1)
    ax.set_yticks([])
    ax.set_facecolor('none')
    ax.spines['polar'].set_color('white')
    ax.grid(linestyle='dotted', color='#bbbbbb', linewidth=0.5)

    wrapped_labels = ['\n'.join(wrap(l, 15)) for l in categories]
    ax.set_xticks(angles[:-1])
    ax.set_xticklabels(wrapped_labels, fontsize=16, color='white', fontproperties=font.copy().set_weight('bold'))
    ax.tick_params(axis='x', pad=35, colors='white')

    progress = min(1.0, frame / 60)
    current_values = [p * progress for p in percentages]
    current_values_loop = current_values + current_values[:1]

    for i in range(N):
        thetas_wedge = [angles[i], angles[i+1], angles[i]]
        radii_wedge = [current_values_loop[i], current_values_loop[i+1], 0]
        ax.fill(thetas_wedge, radii_wedge, color=colors[i], alpha=0.6)
    
    ax.plot(angles, current_values_loop, color='white', linewidth=1.5)

    if frame >= 60:
        for angle, value in zip(angles[:-1], percentages):
            ax.text(
                angle,
                value + 8,
                f'{value:.1f}%',
                ha='center',
                va='center',
                color='white',
                fontsize=13,
                fontproperties=font.copy().set_weight('bold')
            )

ani = FuncAnimation(
    fig, animate, 
    frames=160, 
    interval=40, 
    repeat=True
)

gif_path = "gifs/"
gif_name = "key_benefits"
output_path = gif_path + gif_name + ".gif"

ani.save(
    output_path,
    writer='pillow',
    dpi=500,
    savefig_kwargs={'transparent': True, 'bbox_inches': 'tight', 'pad_inches': 0.2}
)
plt.close(fig)
print(f"✅ GIF '{gif_name.upper()}' exported successfully.")
```
