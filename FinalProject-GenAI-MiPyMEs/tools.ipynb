```python
import matplotlib.pyplot as plt
from matplotlib import font_manager, patches

# Fuente personalizada
font_path = "/Users/<YOUR_USER>/Library/Fonts/AlbertSans-Regular.ttf"
font = font_manager.FontProperties(fname=font_path)

# Datos
tools = ["ChatGPT", "DALL·E", "Copy.ai"]
applications = [
    "Chatbot / Generación de texto",
    "Generación de imágenes",
    "Generación de texto (marketing)"
]
uses = [
    "Responde preguntas de clientes y redacta contenido",
    "Crea logos, diseños o imágenes promocionales",
    "Escribe copys publicitarios y publicaciones"
]
benefits = [
    "Atención 24/7 sin personal extra",
    "Diseño visual económico",
    "Marketing consistente y rápido"
]
colors = ['#c340ff', '#70e0ff', '#5f5ff6']

# Configurar figura
fig, ax = plt.subplots(figsize=(10, 6))
ax.set_xlim(0, 10)
ax.set_ylim(0, 10)
ax.axis('off')

# Distribución vertical
spacing_y = 3

for i in range(3):
    y = 9 - i * spacing_y

    # Pictograma circular
    circle = patches.Circle((1.3, y), 0.4, color=colors[i])
    ax.add_patch(circle)

    # Nombre de la herramienta
    ax.text(2.0, y + 0.6, tools[i],
            fontsize=16, fontproperties=font, color='white', weight='bold', va='top')

    # Aplicación
    ax.text(2.0, y + 0.2, f"Aplicación: {applications[i]}",
            fontsize=12, fontproperties=font, color='white', va='top')

    # Uso
    ax.text(2.0, y - 0.2, f"Uso: {uses[i]}",
            fontsize=11, fontproperties=font, color='white', va='top')

    # Beneficio
    ax.text(2.0, y - 0.6, f"Beneficio: {benefits[i]}",
            fontsize=11, fontproperties=font, color='white', va='top')

# Fondo oscuro y guardado
fig.patch.set_facecolor('black')
plt.savefig("gifs/herramientas_pictografico.png", dpi=300, transparent=True)
plt.close(fig)
print("✅ Gráfico 'herramientas_pictografico.gif' exportado exitosamente.")
```
