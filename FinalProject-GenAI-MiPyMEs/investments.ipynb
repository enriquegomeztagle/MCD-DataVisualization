```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from matplotlib import font_manager
import matplotlib.patheffects as pe
from matplotlib.collections import PolyCollection
from textwrap import wrap

font_path = "/Users/<YOUR_USER>/Library/Fonts/AlbertSans-Regular.ttf"
font = font_manager.FontProperties(fname=font_path)

labels = [
    "Inversión 2023",
    "Intención 2024",
    "No usuarias -> IA generativa",
    "Usuarias -> ampliar IA generativa",
    "Inversión continua 2025"
]
values = np.array([28.0, 73.0, 78.6, 93.4, 70.0])

x_pos = np.arange(len(labels))
x_step = np.repeat(x_pos, 2)[1:]
x_step = np.insert(x_step, 0, x_pos[0])
y_step = np.repeat(values, 2)

frames = 60
pause_frames = 80
total_frames = frames + pause_frames

fig, ax = plt.subplots(figsize=(12, 6), facecolor='none')
fig.patch.set_alpha(0.0)

def style_axes():
    ax.set_facecolor('none')
    ax.set_ylim(0, 110)
    ax.set_xlim(-0.5, len(labels) - 0.5)
    ax.set_xticks(x_pos)
    ax.set_xticklabels(
        ['\n'.join(wrap(label, 20)) for label in labels],
        fontproperties=font,
        color='white',
        fontsize=13
    )
    ax.tick_params(colors='white')
    ax.grid(axis='y', linestyle='dotted', linewidth=0.5, color='white')
    ax.set_yticks([])
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.spines['bottom'].set_color('white')
    ax.spines['left'].set_visible(False)
    ax.set_title(
        "Inversión pasada y planes futuros de las PyMEs mexicanas en IA/IA generativa",
        fontproperties=font,
        fontsize=16,
        color='white',
        pad=20
    )
    ax.margins(y=0.1)

style_axes()
fig.tight_layout(pad=1.5)

line, = ax.plot(
    [], [],
    color='#c340ff',
    linewidth=3,
    zorder=2,
    path_effects=[
        pe.Stroke(linewidth=6, foreground='black', alpha=0.3),
        pe.Normal()
    ]
)

dots = ax.scatter(
    [], [],
    s=100,
    facecolors='#c340ff',
    edgecolors='white',
    linewidth=2,
    zorder=3
)

texts = []
for x in x_pos:
    t = ax.text(
        x, 0, '',
        ha='center', va='bottom',
        fontsize=14,
        color='white',
        fontproperties=font,
        weight='bold',
        path_effects=[
            pe.Stroke(linewidth=3, foreground='black', alpha=0.5),
            pe.Normal()
        ],
        visible=False
    )
    texts.append(t)

def init():
    line.set_data([], [])
    dots.set_offsets(np.empty((0, 2)))
    for t in texts:
        t.set_visible(False)
    for coll in list(ax.collections):
        if isinstance(coll, PolyCollection):
            coll.remove()
    return [line, dots] + texts

def animate(frame):
    prog = min(1.0, frame / frames)
    y_anim = y_step * prog
    line.set_data(x_step, y_anim)
    for coll in list(ax.collections):
        if isinstance(coll, PolyCollection):
            coll.remove()
    fill = ax.fill_between(
        x_step, y_anim,
        color='#c340ff',
        alpha=0.15,
        zorder=1
    )
    if frame >= frames:
        dots.set_offsets(np.column_stack([x_pos, values]))
        for t, (x, y) in zip(texts, zip(x_pos, values)):
            t.set_position((x, y + 4))
            t.set_text(f'{y:.1f}%')
            t.set_visible(True)
    return [line, fill, dots] + texts

ani = FuncAnimation(
    fig,
    animate,
    frames=total_frames,
    init_func=init,
    interval=50,
    blit=True,
    repeat=True
)

gif_path = "gifs/"
gif_name = "investments"
output_path = gif_path + gif_name + ".gif"

ani.save(
    output_path,
    writer='pillow',
    dpi=500,
    savefig_kwargs={'transparent': True}
)
plt.close(fig)

print(f"✅ GIF '{gif_name.upper()}' exported successfully.")
```
