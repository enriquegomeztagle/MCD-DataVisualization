```python
import matplotlib.pyplot as plt
import squarify
import numpy as np
from matplotlib.animation import FuncAnimation
from matplotlib import font_manager
from textwrap import wrap

font_path = "/Users/<YOUR_USER>/Library/Fonts/AlbertSans-Regular.ttf"
font = font_manager.FontProperties(fname=font_path)

labels = [
    "Desconocimiento de la IA generativa",
    "Dificultad para identificar casos de uso",
    "No han implementado aún",
    "No es el momento adecuado",
    "No sabe cómo hacerlo",
    "Integración de datos deficiente",
    "Escasez de talento especializado",
    "Necesidad de capacitación"
]
values = [92, 40, 45.3, 44.7, 41.1, 54, 36, 25]
wrapped_labels = ['\n'.join(wrap(lbl, 15)) for lbl in labels]
colors = ['#c340ff', '#3399FF', '#5f5ff6', '#A673E0', '#70a0ff', '#c340ff', '#5f5ff6', '#3399FF']

normed = squarify.normalize_sizes(values, 100, 100)
rects = squarify.squarify(normed, 0, 0, 100, 100)

fig, ax = plt.subplots(figsize=(8, 8), facecolor='none')
fig.patch.set_alpha(0.0)
ax.axis('off')
ax.set_xlim(0, 100)
ax.set_ylim(0, 100)

def animate(frame):
    ax.clear()
    ax.axis('off')
    ax.set_xlim(0, 100)
    ax.set_ylim(0, 100)
    count = min(frame + 1, len(rects))
    for i in range(count):
        r = rects[i]
        ax.add_patch(plt.Rectangle((r['x'], r['y']), r['dx'], r['dy'], color=colors[i], alpha=0.75, ec='white', lw=2))
        ax.text(
            r['x'] + r['dx']/2,
            r['y'] + r['dy']/2,
            f"{wrapped_labels[i]}\n{values[i]:.1f}%",
            va='center', ha='center',
            fontsize=11, color='white',
            fontproperties=font, fontweight='bold'
        )
    ax.set_title(
        "Principales obstáculos en la implementación de IA generativa",
        fontsize=16, color='white',
        fontproperties=font, pad=20
    )
    return []

drawing_frames = len(labels)
interval = 400
pause_frames = int(10000 / interval)
total_frames = drawing_frames + pause_frames

ani = FuncAnimation(fig, animate, frames=total_frames, interval=interval, repeat=True)

gif_path = "gifs/"
gif_name = "obstacles"
output_path = gif_path + gif_name + ".gif"

ani.save(
    output_path,
    writer='pillow',
    dpi=500,
    savefig_kwargs={'transparent': True}
)
plt.close(fig)

print(f"✅ GIF '{gif_name.upper()}' exported successfully.")
```
