```python
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from matplotlib import font_manager
import matplotlib.patches as patches
from textwrap import wrap

font_path = "/Users/<YOUR_USER>/Library/Fonts/AlbertSans-Regular.ttf"
font = font_manager.FontProperties(fname=font_path)

empresas = [
    {"nombre": "Finanzas PyME",       "sector": "Finanzas PyME",  "aplicacion": "IA generativa para agilizar procesos de crédito y mejorar la experiencia del cliente", "resultado": ">90% reducción en tiempo de respuesta para incorporación de crédito"},
    {"nombre": "Banca", "sector": "Banca",         "aplicacion": "Asistente virtual con IA generativa para operaciones bancarias internas y atención",           "resultado": "ROI ~1,250%; reducción del 80% en tiempo de transacción; mayor satisfacción de clientes"},
    {"nombre": "Salud Startup",                "sector": "Salud startup", "aplicacion": "Plataforma con IA generativa para monitoreo personalizado de pacientes",                        "resultado": "Mayor capacidad de atención; respuestas personalizadas; menos complicaciones médicas"},
    {"nombre": "Banca",                 "sector": "Banca",         "aplicacion": "ChatGPT Enterprise para empleados (atención a clientes, trámites, etc.)",                     "resultado": "3,000 empleados capacitados; mejora en atención y agilización de créditos PyME"},
    {"nombre": "Fintra",                 "sector": "Fintech",       "aplicacion": "IA para optimizar atención al cliente y educación financiera de usuarios",                    "resultado": "Respuestas más rápidas; herramientas educativas para acceso financiero"},
]

colors = ['#c340ff', '#0066CC', '#5f5ff6', '#A673E0', '#70a0ff']
n = len(empresas)

fig, ax = plt.subplots(figsize=(12, 4))
fig.patch.set_alpha(0.0)
ax.axis("off")
ax.set_xlim(0, 1)
ax.set_ylim(0, 1)

pad = 0.05
card_width = (1 - pad * (n + 1)) / n
card_height = 0.8
y0 = 0.1

frames_per_card = 20
total_frames = frames_per_card * n

def draw_card(ax, data, idx, color):
    x0 = pad + idx * (card_width + pad)
    rect = patches.FancyBboxPatch(
        (x0, y0), card_width, card_height,
        boxstyle="round,pad=0.02",
        edgecolor=color, facecolor=color, linewidth=1.5, alpha=0.7,
        transform=ax.transAxes
    )
    ax.add_patch(rect)

    title_lines = wrap(f"{data['nombre']} ({data['sector']})", 18)
    for i, line in enumerate(title_lines):
        ax.text(
            x0 + 0.01,
            y0 + card_height - 0.05 - 0.04 * i,
            line,
            fontsize=12, color='white', fontproperties=font, weight='bold',
            transform=ax.transAxes
        )

    applines = wrap(data['aplicacion'], 20)
    ax.text(
        x0 + 0.01, y0 + card_height - 0.15, "Aplicación:",
        fontsize=9, color='white', fontproperties=font, style='italic',
        transform=ax.transAxes
    )
    for i, line in enumerate(applines):
        ax.text(
            x0 + 0.01,
            y0 + card_height - 0.19 - 0.035 * i,
            line,
            fontsize=9, color='white', fontproperties=font,
            transform=ax.transAxes
        )

    res_start_y = y0 + 0.20
    ax.text(
        x0 + 0.01, res_start_y, "Resultado:",
        fontsize=9, color='white', fontproperties=font, style='italic',
        transform=ax.transAxes
    )
    reslines = wrap(data['resultado'], 20)
    for i, line in enumerate(reslines):
        ax.text(
            x0 + 0.01,
            res_start_y - 0.04 - 0.035 * i,
            line,
            fontsize=9, color='white', fontproperties=font,
            transform=ax.transAxes
        )

def animate(frame):
    ax.clear()
    ax.axis("off")
    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)
    max_idx = min(frame // frames_per_card + 1, n)
    for i in range(max_idx):
        draw_card(ax, empresas[i], i, colors[i % len(colors)])
    return []

ani = FuncAnimation(fig, animate, frames=total_frames, interval=60, repeat=True)

gif_path = "gifs/"
gif_name = "cases"
output_path = gif_path + gif_name + ".gif"

ani.save(
    output_path,
    writer='pillow',
    dpi=500,
    savefig_kwargs={'transparent': True}
)
plt.close(fig)

print(f"✅ GIF '{gif_name.upper()}' exported successfully.")
```
