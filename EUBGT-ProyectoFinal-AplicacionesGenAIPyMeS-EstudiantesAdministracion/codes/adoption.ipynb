```python
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from matplotlib import font_manager

font_path = "/Users/<YOUR_USER>/Library/Fonts/"
font_name="AlbertSans-Regular.ttf"

full_font_path = font_path + font_name

font = font_manager.FontProperties(
    fname=full_font_path
)

categories = ['IA en General', 'IA Generativa']
values = [57.0, 54.7]
color_1 = '#52297b'
color_2 = '#70e0ff'
white = '#ffffff'

fig, ax = plt.subplots(facecolor='none')
fig.patch.set_alpha(0.0)

pause = 100
animation = 60
total = animation + pause

def animate(frame):
    ax.clear()
    ax.set_facecolor('none')
    ax.set_ylim(0, 100)
    for spine in ax.spines.values():
        spine.set_color(white)
    ax.tick_params(colors=white, labelsize=12)
    for lbl in ax.get_xticklabels() + ax.get_yticklabels():
        lbl.set_color(white)
        lbl.set_fontproperties(font)
    if frame < 30:
        h1, h2 = values[0] * (frame/30), 0
    elif frame < 60:
        h1 = values[0]
        h2 = values[1] * ((frame-30)/30)
    else:
        h1, h2 = values
    bars = ax.bar(categories, [h1, h2], color=[color_1, color_2])
    ax.set_title(
        'Adopción de IA en PyMEs mexicanas (2024)',
        fontproperties=font, color=white, fontsize=18
    )
    ax.set_ylabel(
        'Porcentaje (%)',
        fontproperties=font, color=white, fontsize=14
    )
    if frame >= 30:
        ax.text(
            bars[0].get_x() + bars[0].get_width()/2,
            h1 + 2,
            f'{h1:.1f}%',
            ha='center',
            va='bottom',
            color=white,
            fontproperties=font,
            fontsize=14
        )
    if frame >= 60:
        ax.text(
            bars[1].get_x() + bars[1].get_width()/2,
            h2 + 2,
            f'{h2:.1f}%',
            ha='center',
            va='bottom',
            color=white,
            fontproperties=font,
            fontsize=14
        )
    return bars

ani = FuncAnimation(
    fig, animate,
    frames=total,
    interval=30,
    repeat=True
)

gif_path = "gifs/"
gif_name = "adoption"

output_path = gif_path + gif_name + ".gif"

ani.save(
    output_path,
    writer='pillow',
    dpi=700,
    savefig_kwargs={'transparent': True}
)
plt.close(fig)
print(f"✅ GIF '{gif_name.upper()}' exported successfully.")
```
