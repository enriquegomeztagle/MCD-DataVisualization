```python
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from matplotlib import font_manager

font_path = "/Users/<YOUR_USER>/Library/Fonts/"
font_name = "AlbertSans-Regular.ttf"
full_font_path = font_path + font_name

font = font_manager.FontProperties(fname=full_font_path)

uses = [
    "Generación de contenido\ncomercial o literario",
    "Síntesis y resumen\nde información",
    "Traducción de textos"
]
percentages = [78.9, 45.6, 38.2]
colors = ['#c340ff', '#70e0ff', '#5f5ff6']

fig, ax = plt.subplots(figsize=(12, 6), facecolor='none')
fig.subplots_adjust(left=0.35)
fig.patch.set_alpha(0.0)

frames = 60
pause_frames = 100
total_frames = frames + pause_frames

def animate(frame):
    ax.clear()
    ax.set_facecolor('none')
    ax.set_xlim(0, 100)
    ax.set_ylim(-1, len(uses))
    ax.invert_yaxis()
    ax.set_xticks([])
    ax.set_yticks(range(len(uses)))
    ax.set_yticklabels(uses, fontsize=15, color="#ffffff", fontproperties=font)

    for spine in ax.spines.values():
        spine.set_color('none')

    if frame < frames:
        progress = frame / frames
        values = [p * progress for p in percentages]
    else:
        values = percentages

    for i, (val, color) in enumerate(zip(values, colors)):
        ax.plot([0, val], [i, i], color=color, lw=2)
        ax.plot(val, i, 'o', color=color, markersize=12)
        if frame >= frames:
            ax.text(val + 2, i, f'{percentages[i]:.1f}%',
                    va='center', ha='left', color="#ffffff", fontsize=14, fontproperties=font)

    ax.set_title(
        "Principales usos de la IA generativa en PyMEs de México",
        fontsize=18,
        color="#ffffff",
        fontproperties=font
    )

    return fig.artists

ani = FuncAnimation(
    fig, animate,
    frames=total_frames,
    interval=40,
    repeat=True
)

gif_path = "gifs/"
gif_name = "main_use_cases"
output_path = gif_path + gif_name + ".gif"

ani.save(
    output_path,
    writer='pillow',
    dpi=500,
    savefig_kwargs={'transparent': True}
)
plt.close(fig)
print(f"✅ GIF '{gif_name.upper()}' exported successfully with Albert Sans font.")
```
